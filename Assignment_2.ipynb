```python
roll = 1024170073
roll_no= str(roll)
L = [int(digit) * 10 for digit in roll_no] #list comprehension
print(L)
L.append(100)
print(L)
L.insert(2, 50)
print(L)
L.remove(50)
print(L)
L.pop()
print(L)
L.sort()
print(L)
L.sort(reverse=True)
print(L)
print(L[:3])
print(L[-3:])

average = sum(L) / len(L)
new_list = [x for x in L if x > average]
print(new_list)
```

```python
import traceback

scores = tuple(L[:8])
print(f"Scores tuple: {scores}")

high_score = max(scores)
high_index = scores.index(high_score)
low_score = min(scores)
low_count = scores.count(low_score)
print(f"Highest score: {high_score} at index {high_index}")
print(f"Lowest score: {low_score} appears {low_count} times")

reversed_scores = list(scores[::-1])
print(f"Reversed scores as list: {reversed_scores}")

try:
    user_val = int(input("Enter a score to find: "))
    if user_val in scores:
        print(f"First occurrence of {user_val} is at index {scores.index(user_val)}")
    else:
        print(f"{user_val} is not present in the tuple")
except ValueError as e:
  print(f"Caught expected error")
try:
    scores[0] = 100
except TypeError as e:
    print(f"Caught expected error: {e}")
first_score, second_score, *remaining_scores = scores
print(f"First: {first_score}, Second: {second_score}, Remaining: {remaining_scores}")
```
